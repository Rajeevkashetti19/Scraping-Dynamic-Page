```python
import pandas as pd

def filter_dataframe_by_common_states(df, states_list):
    """Filters a DataFrame based on common states with a given list.

    Args:
        df: The DataFrame to filter.
        states_list: The list of states to compare against.

    Returns:
        A filtered DataFrame containing only rows with states
        present in the states_list.
    """

    # Assuming your DataFrame has a column named 'States' containing lists of states
    # Convert the 'States' column to a string representation of lists
    df['States_str'] = df['name'].astype(str)

    # Filter rows where the string representation of the list contains any of the states in states_list
    filtered_df = df[df['States_str'].apply(lambda x: any(state == x for state in states_list))]

    # Drop the temporary 'States_str' column
    filtered_df = filtered_df.drop(columns=['States_str'])

    return filtered_df

# Example usage:
# Assuming your list of states is called 'my_states_list'
# And your second DataFrame is called 'df2'

filtered_df2 = filter_dataframe_by_common_states(df2, states)
print(filtered_df2)
```
